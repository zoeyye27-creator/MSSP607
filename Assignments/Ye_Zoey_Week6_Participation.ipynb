```python
import pandas as pd

df = pd.read_csv("/content/Student Performance Data.csv")

display(df.head())
```

```python
SCORES = ["math score", "reading score", "writing score"]

add_average      = lambda df: df.assign(**{"average score": df[SCORES].mean(axis=1).round(2)})
high_achievers   = lambda df, threshold=240: df[df["total score"] >= threshold].reset_index(drop=True)
to_letter        = lambda avg: "A" if avg >= 90 else "B" if avg >= 80 else "C" if avg >= 70 else "D" if avg >= 60 else "F"
add_letter_grade = lambda df: add_average(df).assign(**{"letter grade": lambda d: d["average score"].apply(to_letter)})
summary_by       = lambda df, col: (df.groupby(col)[SCORES + ["total score"]].mean().round(2)
                                      .reset_index().sort_values("total score", ascending=False))
prep_comparison  = lambda df: summary_by(df, "test preparation course")
```

```python
df_avg = add_average(df)
display(df_avg.head())
```

```python
df_top = high_achievers(df)
print(len(df_top), "students scored 240+")
display(df_top.head())
```

```python
df_grades = add_letter_grade(df)
display(df_grades[["average score", "letter grade"]].head(10))
display(df_grades["letter grade"].value_counts().sort_index())
```

```python
display(summary_by(df, "parental level of education"))
```

```python
display(prep_comparison(df))
```
