```python
import numpy as np

analysis_df = df.copy()

analysis_df["EDU_FATHER"] = analysis_df["EDU_FATHER"].replace({
    "0": np.nan,
    0: np.nan,
    "Ninguno": "No education",
    "Not sure": np.nan
})

analysis_df["EDU_MOTHER"] = analysis_df["EDU_MOTHER"].replace({
    "0": np.nan,
    0: np.nan,
    "Ninguno": "No education",
    "Not sure": np.nan
})

analysis_df["STRATUM"] = analysis_df["STRATUM"].replace({
    "0": np.nan,
    0: np.nan
})
```

```python
def group_summary(data, group_column, score_column="G_SC"):
    return (
        data.groupby(group_column)[score_column]
        .agg(["count", "mean", "median", "std"])
        .sort_values("mean", ascending=False)
        .round(2)
    )
```

```python
from scipy.stats import ttest_ind

male_scores = analysis_df.loc[
    analysis_df["GENDER"] == "M", "G_SC"
].dropna()

female_scores = analysis_df.loc[
    analysis_df["GENDER"] == "F", "G_SC"
].dropna()

t_statistic, p_value = ttest_ind(
    male_scores,
    female_scores,
    equal_var=False
)

print("Mean difference:", round(male_scores.mean() - female_scores.mean(), 2))
print("T-statistic:", round(t_statistic, 2))
print("P-value:", p_value)
```

```python
pooled_standard_deviation = np.sqrt(
    (
        (len(male_scores) - 1) * male_scores.std() ** 2
        + (len(female_scores) - 1) * female_scores.std() ** 2
    )
    / (len(male_scores) + len(female_scores) - 2)
)

# Calculate Cohen's d
cohens_d = (
    male_scores.mean() - female_scores.mean()
) / pooled_standard_deviation

print("Cohen's d:", round(cohens_d, 3))
```

```python
from scipy.stats import f_oneway

stratum_order = [
    "Stratum 1",
    "Stratum 2",
    "Stratum 3",
    "Stratum 4",
    "Stratum 5",
    "Stratum 6"
]

stratum_groups = [
    analysis_df.loc[
        analysis_df["STRATUM"] == stratum, "G_SC"
    ].dropna()
    for stratum in stratum_order
]

f_statistic, p_value = f_oneway(*stratum_groups)

print("F-statistic:", round(f_statistic, 2))
print("P-value:", p_value)
```

```python
overall_mean = analysis_df.loc[
    analysis_df["STRATUM"].isin(stratum_order), "G_SC"
].mean()

ss_between = sum(
    len(group) * (group.mean() - overall_mean) ** 2
    for group in stratum_groups
)

valid_stratum_data = analysis_df.loc[
    analysis_df["STRATUM"].isin(stratum_order), "G_SC"
].dropna()

ss_total = sum((valid_stratum_data - overall_mean) ** 2)

eta_squared = ss_between / ss_total

print("Eta squared:", round(eta_squared, 3))
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

stratum_test_data = analysis_df[
    analysis_df["STRATUM"].isin(stratum_order)
].dropna(subset=["STRATUM", "G_SC"])

tukey_stratum = pairwise_tukeyhsd(
    endog=stratum_test_data["G_SC"],
    groups=stratum_test_data["STRATUM"],
    alpha=0.05
)

print(tukey_stratum)
```

```python
from scipy.stats import f_oneway

def anova_with_eta_squared(data, group_column, score_column="G_SC"):
    clean_data = data.dropna(subset=[group_column, score_column])

    groups = [
        group[score_column].values
        for _, group in clean_data.groupby(group_column)
    ]

    f_statistic, p_value = f_oneway(*groups)

    overall_mean = clean_data[score_column].mean()

    ss_between = sum(
        len(group) * (group[score_column].mean() - overall_mean) ** 2
        for _, group in clean_data.groupby(group_column)
    )

    ss_total = sum(
        (clean_data[score_column] - overall_mean) ** 2
    )

    eta_squared = ss_between / ss_total

    return f_statistic, p_value, eta_squared


father_f, father_p, father_eta = anova_with_eta_squared(
    analysis_df,
    "EDU_FATHER"
)

mother_f, mother_p, mother_eta = anova_with_eta_squared(
    analysis_df,
    "EDU_MOTHER"
)

print("Father's education:")
print("F-statistic:", round(father_f, 2))
print("P-value:", father_p)
print("Eta squared:", round(father_eta, 3))

print("\nMother's education:")
print("F-statistic:", round(mother_f, 2))
print("P-value:", mother_p)
print("Eta squared:", round(mother_eta, 3))
```

```python
program_summary = (
    analysis_df.groupby("ACADEMIC_PROGRAM")["G_SC"]
    .agg(["count", "mean", "median", "std"])
    .sort_values("mean", ascending=False)
    .round(2)
)

display(program_summary)
```

```python
reliable_programs = (
    analysis_df["ACADEMIC_PROGRAM"]
    .value_counts()
    .loc[lambda counts: counts >= 30]
    .index
)

program_test_data = analysis_df[
    analysis_df["ACADEMIC_PROGRAM"].isin(reliable_programs)
].copy()

program_f, program_p, program_eta = anova_with_eta_squared(
    program_test_data,
    "ACADEMIC_PROGRAM"
)

print("Academic program:")
print("Number of programs:", len(reliable_programs))
print("F-statistic:", round(program_f, 2))
print("P-value:", program_p)
print("Eta squared:", round(program_eta, 3))
```

```python
def create_category_summary(data, category_column, category_type):
    summary = (
        data.groupby(category_column)
        .agg(
            n=("G_SC", "count"),
            G_SC_mean=("G_SC", "mean"),
            G_SC_std=("G_SC", "std"),
            QR_PRO_mean=("QR_PRO", "mean"),
            QR_PRO_std=("QR_PRO", "std"),
            CR_PRO_mean=("CR_PRO", "mean"),
            CR_PRO_std=("CR_PRO", "std")
        )
        .reset_index()
        .rename(columns={category_column: "Category"})
    )

    summary.insert(0, "Category_Type", category_type)
    return summary


gender_numeric_summary = create_category_summary(
    analysis_df,
    "GENDER",
    "Gender"
)

gender_numeric_summary["Category"] = (
    gender_numeric_summary["Category"]
    .replace({
        "F": "Female",
        "M": "Male"
    })
)

stratum_numeric_summary = create_category_summary(
    analysis_df[
        analysis_df["STRATUM"].isin(stratum_order)
    ],
    "STRATUM",
    "Socioeconomic stratum"
)

stratum_numeric_summary["Category"] = pd.Categorical(
    stratum_numeric_summary["Category"],
    categories=stratum_order,
    ordered=True
)

stratum_numeric_summary = stratum_numeric_summary.sort_values(
    "Category"
)

category_numeric_summary = pd.concat(
    [
        gender_numeric_summary,
        stratum_numeric_summary
    ],
    ignore_index=True
)

numeric_columns = category_numeric_summary.columns.difference(
    ["Category_Type", "Category", "n"]
)

category_numeric_summary[numeric_columns] = (
    category_numeric_summary[numeric_columns].round(2)
)

display(category_numeric_summary)
```
