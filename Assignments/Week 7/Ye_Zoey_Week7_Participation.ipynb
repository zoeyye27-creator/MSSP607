```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv(
    '/content/MSSP607/data/InternetSales.csv',
    encoding='cp1252',
    low_memory=False
)
data['OrderDate'] = pd.to_datetime(data['OrderDate'])
data['Year'] = data['OrderDate'].dt.year

yearly_sales = data.groupby('Year')['SalesAmount'].sum()
yearly_sales
```

```python
fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(1, 1, 1)

yearly_sales.plot.bar(ax=ax, color='b', alpha=0.7, rot=0)

ax.set_title('Yearly Internet Sales')
ax.set_xlabel('Year')
ax.set_ylabel('Total Sales')

ax.ticklabel_format(style='plain', axis='y')

plt.tight_layout()

fig.savefig('/content/YearlySales.pdf')

plt.show()
```
