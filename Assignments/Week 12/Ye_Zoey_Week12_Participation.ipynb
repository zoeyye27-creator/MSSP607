```python
import pandas as pd
import numpy as np
import scipy.stats as stats

data = pd.read_csv('/content/MSSP607/Assignments/MSSP607/data/Student Performance Data.csv')
data
```

H0: Male students in the two groups have the same mean total score.

H1: Male students in the two groups have different mean total scores.

```python
male = data[data['gender'] == 'male']

completed = male[male['test preparation course'] == 'completed']['total score']
none = male[male['test preparation course'] == 'none']['total score']

print('Completed mean:', np.mean(completed))
print('None mean:', np.mean(none))

n1 = len(completed)
n2 = len(none)

variance1 = stats.sem(completed)**2 * n1
variance2 = stats.sem(none)**2 * n2

pooled_variance = ((n1 - 1)*variance1 + (n2 - 1)*variance2)/(n1 + n2 - 2)

tval = (np.mean(completed) - np.mean(none))/(pooled_variance*(1/n1 + 1/n2))**0.5
print('t-value:', tval)

if tval < 0:
    tval = -tval

td = stats.t(n1 + n2 - 2)
p = 2*td.sf(tval)
print('p-value:', p)

alpha = 0.05
if p < alpha:
    print('Reject H0. The mean total scores are significantly different.')
else:
    print('Fail to reject H0. The difference is not statistically significant.')
```
