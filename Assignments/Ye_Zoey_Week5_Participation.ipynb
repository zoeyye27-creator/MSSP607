```python
from pathlib import Path
import pandas as pd

data_dir = Path("/content/MSSP607/data")
tickers = ["fb", "aapl", "amzn", "nflx", "goog"]
```

```python
frames = []

for ticker in tickers:
    df = pd.read_csv(data_dir / f"{ticker}.csv")
    df["ticker"] = ticker
    frames.append(df)
```

```python
stocks = pd.concat(frames, ignore_index=True)
```

```python
stocks.to_csv(data_dir / "faanf.csv", index=False)
stocks.head()
```

```python
stocks = pd.read_csv(data_dir / "faanf.csv")

stocks["date"] = pd.to_datetime(stocks["date"])
stocks["volume"] = pd.to_numeric(stocks["volume"]).astype(int)

stocks = stocks.sort_values(["date", "ticker"])
print(stocks[["date", "ticker", "volume"]].head())
```

```python
top_seven = stocks.nlargest(7, "volume")
top_seven
```

```python
melted = stocks.melt(
    id_vars=["date", "ticker"],
    var_name="variable",
    value_name="value"
)
melted.head()
```
