```python
from google.colab import userdata
import os
import subprocess

github_token = userdata.get('Git_Key')
owner = 'zoeyye27-creator'
repository = 'MSSP607'

clone_url = f'https://x-access-token:{github_token}@github.com/{owner}/{repository}.git'

subprocess.run(['git', 'clone', clone_url], check=True)

os.chdir(os.path.join(repository, 'Assignments'))
print(f"Changed directory to: {os.getcwd()}")
```

```python
import requests, bs4, re

category = 'cats'
url = 'https://www.flickr.com/search/?text=' + category

res = requests.get(url)
res.raise_for_status()
soup = bs4.BeautifulSoup(res.text, 'html.parser')
```

```python
images = soup.find_all('img', {'src': re.compile(r'live\.staticflickr\.com/')})

number = 1
for image in images:
    image_url = image['src']
    if image_url[:2] == '//':
        image_url = 'https:' + image_url

    try:
        res = requests.get(image_url)
        res.raise_for_status()
    except Exception as exc:
        print('There was a problem: %s' % (exc))
    else:
        filename = 'image_%s.jpg' % (number)
        image_file = open(filename, 'wb')
        for chunk in res.iter_content(100000):
            image_file.write(chunk)
        image_file.close()
        print('Downloaded: %s' % (filename))
        number = number + 1

print('Downloaded %s images.' % (number - 1))
```

```python
github_token = userdata.get('Git_Key')

remote_url = (
    f'https://x-access-token:{github_token}'
    f'@github.com/{owner}/{repository}.git'
)

result = subprocess.run(
    ['git', 'remote', 'set-url', 'origin', remote_url],
    capture_output=True,
    text=True
)

if result.returncode == 0:
    result = subprocess.run(
        ['git', 'push', 'origin', 'HEAD'],
        capture_output=True,
        text=True
    )

message = (result.stdout + result.stderr).replace(
    github_token, '[TOKEN HIDDEN]'
)
print(message)

if result.returncode == 0:
    print('Images uploaded to GitHub.')
```
